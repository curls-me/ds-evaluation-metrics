#### Mean Absolute Error (MAE)

The [mean absolute error](https://scikit-learn.org/stable/modules/generated/sklearn.metrics.mean_absolute_error.html) (or MAE) is the average of the absolute differences between predictions and actual values. It gives an idea of how wrong the predictions were. The measure gives an idea of the magnitude of the error, but no idea of the direction (e.g. over or under predicting).

```python
# Your code for MAE!
def mae_function(y_true, y_pred):
    absolut_error = np.abs(y_true - y_pred)
    return np.mean(absolut_error)
```

#### Mean Squared Error (MSE)

The [mean squared error](https://scikit-learn.org/stable/modules/generated/sklearn.metrics.mean_squared_error.html) (or MSE) is much like the mean absolute error in that it provides a gross idea of the magnitude of error.
Taking the square root of the mean squared error converts the units back to the original units of the output variable and can be meaningful for description and presentation. This is called the **root mean squared error** (or RMSE).

```python
# Your code for MSE (or RMSE)!
def mse_function(y_true, y_pred, root=False):
    squared_error = np.power((y_true - y_pred), 2)
    if root:
        return np.sqrt(np.mean(squared_error))
    else:
        return np.mean(squared_error)
```

#### R-squared

The [R^2](https://scikit-learn.org/stable/modules/generated/sklearn.metrics.r2_score.html) (or R-squared) metric provides an indication of the goodness of fit of a set of predictions to the actual values. In statistical literature, this measure is called the coefficient of determination.
A perfect prediction has an R-squared score of 1. A score of 0 matches a model that always predicts the mean of the true values; a negative score means the predictions are worse than that baseline.

```python
# Your code for R-squared!
def r2_function(y_true, y_pred):
    ssr = np.sum(np.power((y_true - y_pred), 2))
    sst = np.sum(np.power((y_true - np.mean(y_true)), 2))
    return 1 - (ssr / sst)
```

#### Test your functions
Start with the import of the necessary functions and modules. 

```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score,
    root_mean_squared_error,
)
from sklearn.model_selection import train_test_split
```

```python
# Import data
df = pd.read_csv("../data/wine-quality.csv", sep=";")
df.head(2)
```

```python
# Define features and target
X = df.drop("alcohol", axis=1)
y = df.alcohol
```

```python
# Train test split
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)
```

```python
# Fit model on train data
linreg = LinearRegression()
linreg.fit(X_train, y_train)

# Make prediction on test data
y_pred = linreg.predict(X_test)
```

### Evaluate the model

```python
# MAE own function
print("MAE with self-implemented function: ", mae_function(y_test, y_pred))

# MAE scikit-learn function

print("    MAE with scikit-learn function: ", mean_absolute_error(y_test, y_pred))
```

```python
# MSE own function
print(" MSE with self-implemented function: ", mse_function(y_test, y_pred))
print("RMSE with self-implemented function: ", mse_function(y_test, y_pred, root=True))

# MSE scikit-learn function
print("     MSE with scikit-learn function: ", mean_squared_error(y_test, y_pred))
print("    RMSE with scikit-learn function: ", root_mean_squared_error(y_test, y_pred))
```

```python
# R-squared own function
print("R-squared with self-implemented function: ", r2_function(y_test, y_pred))

# R-squared scikit-learn function
print("    R-squared with scikit-learn function: ", r2_score(y_test, y_pred))
```
