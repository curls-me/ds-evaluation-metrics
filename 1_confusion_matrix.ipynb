The following code demonstrates how to create a confusion matrix on a predicted model.


For this, we have to import the `confusion_matrix` module from the `sklearn.metrics` library which helps us to generate the confusion matrix.

```python
from sklearn.metrics import confusion_matrix
```

```python
# Within a confusion matrix, we are comparing the actual target values (y_test) with the values predicted by our model (often called y_pred)
# Here, we are creating random values for our actual and predicted values as example
actual = [1, 1, 0, 1, 0, 0, 1, 0, 0, 0]
predicted = [1, 0, 0, 1, 0, 0, 1, 1, 1, 0]

# Here, we are creating a confusion matrix which compares actual and predicted values
results = confusion_matrix(actual, predicted)

print("Confusion Matrix :\n", results)
```

## Precision vs Recall: Choosing the Right Trade-off

A classic Data Science interview question is:

> *Which is better: more false positives or more false negatives?*

This is a trick question because there is no single correct answer.

The real answer is:

> *It depends on the problem.*

Different applications assign different costs to different types of errors. In some situations, **false positives** are more costly; in others, **false negatives** have more serious consequences.

Consider **credit card fraud detection** as an example.

* A **False Positive** occurs when a legitimate transaction is incorrectly flagged as fraudulent. This may inconvenience or frustrate the customer.
* A **False Negative** occurs when a fraudulent transaction is incorrectly classified as legitimate. In this case, the company may need to reimburse the customer and absorb the financial loss.

Although neither type of error is ideal, missing fraudulent transactions is typically more costly than incorrectly flagging legitimate ones.

This trade-off directly influences how we evaluate classification models. Some metrics place greater emphasis on **reducing false positives**, while others focus more on **reducing false negatives**.

Two of the most common metrics used to evaluate these trade-offs are **precision** and **recall**, which we will introduce next.
